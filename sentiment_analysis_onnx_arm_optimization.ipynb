```python
from transformers import pipeline
import time

classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

start = time.time()
result = classifier("I love this hackathon, it's amazing!")
end = time.time()

print(result)
print(f"Processing time: {end-start:.4f} seconds")
```

```python
from optimum.onnxruntime import ORTModelForSequenceClassification
from transformers import AutoTokenizer

model = ORTModelForSequenceClassification.from_pretrained(
    "distilbert-base-uncased-finetuned-sst-2-english",
    export=True
)
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")

model.save_pretrained("optimized_model")
tokenizer.save_pretrained("optimized_model")

print("Model successfully optimized and saved!")
```

```python
import time
from optimum.pipelines import pipeline as ort_pipeline

# Test original model
start = time.time()
result1 = classifier("Testing speed comparison")
time1 = time.time() - start

# Test optimized model (ONNX)
optimized_classifier = ort_pipeline("text-classification", model="optimized_model", tokenizer="optimized_model")

start = time.time()
result2 = optimized_classifier("Testing speed comparison")
time2 = time.time() - start

print(f"Original model: {time1:.4f} seconds")
print(f"Optimized model (ONNX): {time2:.4f} seconds")
print(f"Improvement: {((time1-time2)/time1*100):.1f}%")
```
